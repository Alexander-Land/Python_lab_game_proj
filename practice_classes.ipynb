```python
# alexs code
import math
import random


class spells(object):
    """Creates spells for wizard
    spells have a damage value and an element"""
    def __init__(self,name,damage,element):
        self.name = name
        self.damage = damage
        self.element = element

    def attack(self, enemy):
        """deals damage to enemy depending on elemental weakness"""
        if self.element == enemy.weakness:
            enemy.health -= self.damage*1.5
        else:
            enemy.health -= self.damage

class enemy(object):
    """creates enemy"""
    def __init__(self,name,health,weakness,damage,speed,speech=None):
        self.name = name
        self.health = health
        self.weakness = weakness
        self.damage = damage
        self.speed = speed
        self.speech = speech

    def attack(self,player):
        """attacks player, speed calculations will be done elsewhere"""
        player.health -= self.damage

    def speak(self):
        """prints the saved line that the creature would say"""
        print("\"", self.speech,"\"")


#maggies code 

class weapons(object):
    """creates weapon for warrior"""
    def __init__(self, name, damage, speed):
        self.name = name
        self.damage = damage
        self.speed = speed

    def attack(self, enemy):
        enemy.health -= self.damage
        
        speed_diff = self.speed - enemy.speed
        if speed_diff > 0:
            probability_of_second_hit = 1 / (1 + math.exp((3-speed_diff)))
            if probability_of_second_hit*100 > random.randint(1,100):
                enemy.health -= self.damage
        # if speed difference is great enough the weapon has a chance to hit twice.


class player(object):
    """creates player object"""
    def init(self, name, health, damage):
        self.name = name
        self.health = health
        self.damage = damage



class consumable(object):
    def __init__(self):
        pass

    def useitem():
        pass
    

class inventory(object):
    """holds items"""
    def __init__(self):
        self.inventory_list = []

    def create_item(self, item, quantity):
        """adds an item to the inventory list and guarantees a certain number is in that list"""
        self.inventory_list.append({'item': item, 'quantity': quantity})

    def add_item(self,item, quantity):
        """increments the number of items, if item doesn't exists it creates the item"""
        self.item_found = False

        for entry in self.inventory_list:
            if entry['item'] == item:
                entry['quantity'] += quantity  # Increment the quantity of the existing item
                self.item_found = True
                break
        
        # If the item doesn't exist, create a new item with the given quantity
        if not self.item_found:
            self.create_item(item, quantity)


    def use_item(self, item):
        """Decrements the quantity of an item by 1 if it exists and has quantity > 0. 
        removes the item if quantity reaches 0.
        if item exists and is an item class it'll use it.
        this section is chatgpt code
        """
        for entry in self.inventory_list:
            if entry['item'] == item:
                if entry['quantity'] > 0:
                    
                    if isinstance(entry['item'], consumable): 
                        consumable.useitem()

                    entry['quantity'] -= 1  # Decrease the quantity by 1
                    print(f"Used 1 {item}. Remaining: {entry['quantity']}")

                    # If the quantity reaches 0, remove the item from the list
                    if entry['quantity'] == 0:
                        self.inventory_list.remove(entry)
                        print(f"{item} has been removed from inventory as its quantity reached 0.")
                    return
                
                else:
                    print(f"Cannot use {item}.")
                    return
        
        print(f"{item} not found in inventory.")
                
#maggie's code               
class map(self):
    """creates a map of the game"""
    def __init__(self):
        self.rooms = {}
        self.current_room = 'Start'
    rooms = {'Start': {'East': 'The Drowned Gate', #Home of the Game
                       'South': 'The Embered Threshold',
                       'West': 'Ancient Convergence',
                       'Item': {'name' : 'Rusty Sword',
                                'damage': 5,
                                'speed': 1,
                                'type':'weapon'}}, 
            'The Embered Threshold': {'North': 'Start', #Fire Area
                                       'East': 'Smoldering Ascent',
                                       'South': 'Magma Crucible',
                                       'West': 'Stormfire Bastion',
                                       'Item': {'name':'Fireball',
                                                'damage': 10,
                                                'element':'fire',
                                                'type':'spell'}},
            'Stormfire Bastion': {'East': 'The Embered Threshold',
                                  'Item': {'name': 'Blazing Maelstrom',
                                           'damage': 15,
                                           'element': 'fire',
                                           'type':'spell'}},
            'Magma Crucible': {'North': 'The Embered Threshold',
                               'Item': {'name':'Volcanic Maul',
                                        'damage': 15,
                                        'speed': 1,
                                        'type':'weapon'}},
            'Smoldering Ascent': {'West': 'The Embered Threshold',
                                  'Item': {'name': 'Inferno Fang',
                                           'damage': 10,
                                           'speed': 1,
                                           'type':'weapon'}},
            'The Drowned Gate': {'West': 'Start', #Water Area
                                 'South': 'Frozen Depths',
                                 'East': 'Whirlpool Shrine',
                                 'North': 'Abyssal Cavern',
                                 'Item': 'Water Blast'},
            'Abyssal Cavern': {'South': 'The Drowned Gate',
                               'Item': 'Abyssal Trident'},
            'Frozen Depths': {'North': 'The Drowned Gate',
                              'Item': 'Glacier Fang'},
            'Whirlpool Shrine': {'West': 'The Drowned Gate',
                                 'Item': 'Whirlpool Staff'},
            'Ancient Convergence': {'East': 'Start'} #end of game area, final boss
           
    }
#result of last move 
msg = " "


#maggie/alex code
#gameplay loop
while True:
    clear()

    #intro to game, gives player initial greeting and sets up player
    print (f"Welcome to Rift of Tides and Flames! You are a warrior who has been tasked\n 
           with defeating the evil wizard who has been terrorizing the land. You must travel through\n 
           the rift to find him and defeat him. You will have to fight many enemies along the way, \n
           so be prepared. Good luck!")
    print (f"You are in {current_room}. \n Inventory: {self.inventory_list}\n {'-' * 27}")
    print(msg)





```

```python
# Alex's unit tests
import unittest

class SpellDamage_Test(unittest.TestCase):
    def setUp(self):
        self.Fireball = spells("Fireball", 30, "fire")
        self.waterblast = spells("Waterblast", 30, "water")

        self.dragon = enemy("Dragon", 100, "fire", 20, 10)

    def test_Fireball(self):
        """Fire should deal extra damage because dragon is weak to fire
        30*1.5 = 45 damage, 100 health - 45 damage should leave dragon with 55 health"""
        self.Fireball.attack(self.dragon)
        self.assertEqual(55,self.dragon.health)

        """waterblast is listed also as 30 damage but the dragon is not weak to water so
        no extra damage is done.
        55 remaining health - 30 damage = 25 health after waterblast"""
        self.waterblast.attack(self.dragon)
        self.assertEqual(25,self.dragon.health)


        
            
```
